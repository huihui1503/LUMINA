```python
import json
import random
import re
```

```python
def read_json(file_path):
    data = []

    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            data.append(json.loads(line))

    return data
```

```python
dolly_data = read_json("dolly/merged.jsonl")
dolly_length = len(dolly_data)
data_dict = []
for i in range(dolly_length):
    random_index = random.randrange(dolly_length)
    if random_index == i:
        random_index = int(random_index/2) + 1
    
    correct_passages = dolly_data[i]["source_info"]["passages"].replace("\n", " ")
    correct_passages = correct_passages.replace("  ", " ")

    random_passages = dolly_data[random_index]["source_info"]["passages"].replace("\n", " ")
    random_passages = random_passages.replace("  ", " ")


    correct_instruction = dolly_data[i]["prompt"].replace("\n", " ")
    correct_instruction = correct_instruction.replace("  ", " ")

    random_instruction = correct_instruction.replace(correct_passages, random_passages)

    record = {
        "labels": 1 if len(dolly_data[i]["labels"]) > 0 else 0,
        "model": dolly_data[i]["model"],
        "prompt_w_context": correct_instruction,
        "response": dolly_data[i]["response"],
        "prompt_w_random_context": random_instruction,
        "split": dolly_data[i]["split"],
    }
    data_dict.append(record)
    
with open("dolly/preprocessed_merged.jsonl", "w", encoding="utf-8") as f:
    for item in data_dict:
        f.write(json.dumps(item, ensure_ascii=False) + "\n")
```

```python
hallurag_data = read_json("hallurag/merged.jsonl")
hallurag_length = len(hallurag_data)
data_dict = []

def extract_hallurag_context(prompt):

    if "You are an assistant for question-answering tasks." in prompt:
        match =  re.search(r"Context:(.*?)Answer:", prompt, flags=re.DOTALL)
    elif "Your knowledge is limited to only this information:" in prompt:
        match = re.search(r"Your knowledge is limited to only this information:(.*?)QUESTION:", prompt, flags=re.DOTALL)
    elif "Only use the information included in these chunks to answer the question:" in prompt:
       match = re.search(r"Only use the information included in these chunks to answer the question:(.*?)QUESTION:", prompt, flags=re.DOTALL)
    if match:
        return match.group(1).strip()

for i in range(hallurag_length):
    random_index = random.randrange(hallurag_length)
    if random_index == i:
        random_index = int(random_index/2) + 1

    correct_passages = extract_hallurag_context(hallurag_data[i]['prompt'])
    random_passages = extract_hallurag_context(hallurag_data[random_index]['prompt'])
    
    correct_instruction = hallurag_data[i]['prompt']

    random_instruction = correct_instruction.replace(correct_passages, random_passages)

    record = {
        "labels": 1 if len(hallurag_data[i]["labels"]) > 0 else 0,
        "model": hallurag_data[i]["model"],
        "prompt_w_context": correct_instruction,
        "response": hallurag_data[i]["response"],
        "prompt_w_random_context": random_instruction,
        "split": hallurag_data[i]["split"],
    }
    data_dict.append(record)

with open("hallurag/preprocessed_merged.jsonl", "w", encoding="utf-8") as f:
    for item in data_dict:
        f.write(json.dumps(item, ensure_ascii=False) + "\n")
```

```python
ragtruth_data = read_json("ragtruth/merged.jsonl")
ragtruth_length = len(ragtruth_data)
data_dict = []

def extract_ragtruth_context(data_dict):
    if data_dict["task_type"] == "Summary":
        return data_dict["source_info"]
    elif data_dict["task_type"] == "QA":
        return data_dict["source_info"]["passages"]
    elif data_dict["task_type"] == "Data2txt":
        return str(data_dict["source_info"])
    return None

for i in range(ragtruth_length):
    random_index = random.randrange(ragtruth_length)
    if random_index == i:
        random_index = int(random_index/2) + 1

    correct_passages = extract_ragtruth_context(ragtruth_data[i])
    random_passages = extract_ragtruth_context(ragtruth_data[random_index])
    
    correct_instruction = ragtruth_data[i]['prompt']
    random_instruction = correct_instruction.replace(correct_passages, random_passages)

    record = {
        "labels": 1 if len(ragtruth_data[i]["labels"]) > 0 else 0,
        "model": ragtruth_data[i]["model"],
        "prompt_w_context": correct_instruction,
        "response": ragtruth_data[i]["response"],
        "prompt_w_random_context": random_instruction,
        "split": ragtruth_data[i]["split"],
    }
    data_dict.append(record)

with open("ragtruth/preprocessed_merged.jsonl", "w", encoding="utf-8") as f:
    for item in data_dict:
        f.write(json.dumps(item, ensure_ascii=False) + "\n")
```
